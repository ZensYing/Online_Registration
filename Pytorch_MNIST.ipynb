```
#@title Import Dependencies

import torch # deep learning and GenAI
import torch.nn as nn # change from writing torch.nn ទៅការសរសេរកាត់ nn - neural network
import torchvision.datasets as dsets # ដូរពិការសរសេរ torchvision.datasets ទៅជាការសរសេរកាត់ ទៅជា dset
import torchvision.transforms as transforms #ដូរពិការសរសេរ torchvision.transforms ទៅជាការសរសេរកាត់ transforms
from torch.autograd import Variable #អថេរប្រើដើម្បីដំណើរការ update parameters របស់ AI model
```

```
#@title Define Hyperparameters

input_size = 784 # img_size = (28,28) ---> 28*28=784 in total
hidden_size = 500 # number of nodes at hidden layer
num_classes = 10 # number of output classes discrete range [0,9]
num_epochs = 20 # number of times which the entire dataset is passed throughout the model
batch_size = 100 # the size of input data took for one iteration
lr = 1e-3 # size of step - learning rate = step size
```

```
#@title Downloading MNIST data
#Dataset have 3 = Train Data (using to train), Test Data (using to test), Validation Data (using to check performance during training)

train_data = dsets.MNIST(root = './data', train = True,
                        transform = transforms.ToTensor(), download = True)

test_data = dsets.MNIST(root = './data', train = False,
                       transform = transforms.ToTensor())
```

```
#@title Loading the data
#DataLoader -> ជា Container មួយដើម្បីផ្ទុកនូវ data សម្រាប់ Learning Process in AI of torch framework

train_gen = torch.utils.data.DataLoader(dataset = train_data,
                                             batch_size = batch_size,
                                             shuffle = True)
#Train Gen = Train Generater

test_gen = torch.utils.data.DataLoader(dataset = test_data,
                                      batch_size = batch_size,
                                      shuffle = False)
```

```
#@title Define model class (ANN)

class Net(nn.Module):
  #function to define the initial parameters for AI model
  def __init__(self, input_size, hidden_size, num_classes):
    super(Net,self).__init__()
    #matching layer input to hidden - Fully connection network
    # Create fully connection network using pytorch -> nn.Linear
    # 784 -> 500
    self.fc1 = nn.Linear(input_size, hidden_size)
    #activation function -> ReLU -> ប្រើដើម្បីកំណត់យកតួលេខដែរត្រូវបញ្ជូលទៅដំណាក់បន្ទាប់
    #ReLU = Rectified Linear Unit -> បើសិនលេខតូចជាងសូន្យ ដាក់ស្មើ ០ ហើយបើធំជាងសូន្យ យកទៅដំណាក់បន្ទាប់
    self.relu = nn.ReLU()
    # fc = fully connection function -> 500 ->10
    self.fc2 = nn.Linear(hidden_size, num_classes)

  def forward(self,x):
    out = self.fc1(x)
    out = self.relu(out)
    out = self.fc2(out)
    return out
```

```
#@title Build the model

net = Net(input_size, hidden_size, num_classes)
#Check Whether the system contains of GPU computing
if torch.cuda.is_available():
  # cuda -> computing on GPU/TPU
  net.cuda()
```

```
#@title Define loss-function & optimizer
#អនុគមន៍ត្រូវពិនិត្យ តម្លៃទស្សន៍ទាយ និងតម្លៃពិត
loss_function = nn.CrossEntropyLoss()
#optimizer ប្រើដើម្បី កែតម្រូវ model parameter អោយការទស្សន័ទាយកាន់តែល្អ
optimizer = torch.optim.Adam( net.parameters(), lr=lr)
```

```
#@title Training the model

for epoch in range(num_epochs):
  for i ,(images,labels) in enumerate(train_gen):
    # Variable to store data into torch framework
    # images.view -> transorm shape/dimension to 1 vector or n dimen
    images = Variable(images.view(-1,28*28)).cuda() #dimension (784,1)
    labels = Variable(labels).cuda() # 0,1,2...9

    optimizer.zero_grad() # -> zero_grad -> ដាក់បំលាស់ប្ដូរស្មើសូន្យ មុនដំណើរការ update parameter
    # Predict output from input via model
    outputs = net(images)
    # calculatioin loss
    loss = loss_function(outputs, labels)
    #update parameters
    loss.backward()
    optimizer.step()

    if (i+1) % 100 == 0:
      print('Epoch [%d/%d], Step [%d/%d], Loss: %.4f' %(epoch+1, num_epochs, i+1, len(train_data)//batch_size, loss.item()))
```

```
#@title Evaluating the accuracy of the model

correct = 0
total = 0
for images,labels in test_gen:
  images = Variable(images.view(-1,28*28)).cuda()
  labels = labels.cuda()

  output = net(images)
  _, predicted = torch.max(output,1)
  correct += (predicted == labels).sum()
  total += labels.size(0)

print('Accuracy of the model: %.3f %%' %((100*correct)/(total+1)))
```
